## Filling missing values in temperature column

```python
df['temperature_c'] = df['temperature_c'].fillna(df.groupby('device_id')['temperature_c'].transform('mean'))
```

## Changing data type of timestamp column

```python
df['reading_ts'] = pd.to_datetime(df['reading_ts'], utc = True)
df.dtypes
```

## Making separate date and time columns

```python
df['date'] = df['reading_ts'].dt.date
df['time'] = df['reading_ts'].dt.time

df.head(10)
```

## Loading processed data into MySQL server

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

username = "root"
password = "Password" 
host = "127.0.0.1"
port = 3306
database = "mydatabase"

connection_url = URL.create(
    "mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

with engine.connect() as connection:
    print("Successfully connected to MySQL!")
```

```python
table_name = "sensor_data"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False
)

print("DataFrame successfully saved to MySQL!")
```

```python
result = pd.read_sql(
    "SELECT * FROM sensor_data LIMIT 5;",
    con=engine
)

display(result)
```
